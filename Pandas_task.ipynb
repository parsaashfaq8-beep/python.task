```python
#Task:1
#Create this student DataFrame:
import pandas as pd

data ={
    "Name": ["Ali", "Sara", "Ahmed","Ayesha", "Zain"],
    "Marks" : [70, 85, 60, 95, 45]
}
df = pd.DataFrame(data)
print(df)
Marks=df["Marks"]
print(Marks.mean())
print(Marks.max())
print(Marks.min())



name =df["Name"]
print(name.count())
```

```python
passed_students = df[df["Marks"] >= 50]
print("Passed Students:")
print(passed_students)

failed_students = df[df["Marks"] < 50]
print("Failed Students:")
print(failed_students)
```

```python
df["cities"]=["karachi","lahore","faisalabad","islamabad","rawalpindi"]
print("only karachi students:")
print(df[df['cities'] == "karachi"])
```
