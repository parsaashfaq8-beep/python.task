```python
import numpy as np

marks = [60,80,42,63,35,75,95]

avg_marks = (np.mean(marks))
print(avg_marks)

highest_marks = (np.max(marks))
print(highest_marks)

lowest_marks = (np.min(marks))
print(lowest_marks)

marks = np.array(marks)
passed = marks[marks >= 50]
print(passed)

failed = marks[marks < 50]
print(failed)
```
